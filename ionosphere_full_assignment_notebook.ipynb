## 1. Dataset Description

The Ionosphere dataset contains radar signal data collected from a radar system located in Goose Bay, Labrador.
The system consists of a phased array of 16 high-frequency antennas with a total transmitted power of approximately
6.4 kilowatts.

The radar signals were transmitted into the ionosphere and the returned signals were analyzed to determine the
presence of structures in the ionosphere.

The goal of this dataset is to classify radar returns as either:

- **Good (g)** → showing evidence of ionospheric structure
- **Bad (b)** → signals that pass through the ionosphere without significant reflection

The radar system used 17 pulse numbers and each pulse generated two attributes, resulting in 34 numerical features.

## 2. Import Libraries

```python
import pandas as pd
import numpy as np
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler, LabelEncoder
import matplotlib.pyplot as plt
import seaborn as sns
```

## 3. Load Dataset

```python
url = 'https://archive.ics.uci.edu/ml/machine-learning-databases/ionosphere/ionosphere.data'

df = pd.read_csv(url, header=None)

# Rename last column as class
df.rename(columns={34:'class'}, inplace=True)

df.head()
```

## 4. Basic Dataset Information

```python
df.info()
```

```python
df.describe()
```

```python
df.isnull().sum()
```

## 5. Handling Missing Values

Numerical values are imputed using mean, and categorical values using most frequent method.

```python
# Numerical imputation
num_imputer = SimpleImputer(strategy='mean')
df.iloc[:, :-1] = num_imputer.fit_transform(df.iloc[:, :-1])
```

```python
# Categorical imputation
cat_imputer = SimpleImputer(strategy='most_frequent')
df['class'] = cat_imputer.fit_transform(df[['class']])
```

## 6. Outlier Detection using IQR

```python
Q1 = df.iloc[:, :-1].quantile(0.25)
Q3 = df.iloc[:, :-1].quantile(0.75)
IQR = Q3 - Q1

outliers = ((df.iloc[:, :-1] < (Q1 - 1.5 * IQR)) |
            (df.iloc[:, :-1] > (Q3 + 1.5 * IQR)))

print('Total Outliers:', outliers.sum().sum())
```

## 7. Outlier Visualization

```python
plt.figure(figsize=(10,5))
sns.boxplot(data=df.iloc[:,0:10])
plt.title('Outlier Visualization (First 10 Features)')
plt.show()
```

## 8. Feature Scaling

```python
scaler = StandardScaler()
scaled_features = scaler.fit_transform(df.iloc[:, :-1])
```

## 9. Encoding Target Variable

```python
le = LabelEncoder()
df['class_encoded'] = le.fit_transform(df['class'])

df.head()
```

## 10. Feature Engineering

```python
# Feature 1: Mean signal strength
df['signal_mean'] = df.iloc[:, :10].mean(axis=1)

# Feature 2: Signal variance
df['signal_variance'] = df.iloc[:, :10].var(axis=1)

df.head()
```
