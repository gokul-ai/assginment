# Ionosphere Dataset Preprocessing and Feature Engineering
**Dataset Source:** https://www.kaggle.com/datasets/jbrownlee/ionosphere

This notebook demonstrates loading an open-source ionosphere dataset, handling missing values, detecting outliers,
performing transformations, and creating new features.

## 1. Import Libraries

```python
import pandas as pd
import numpy as np
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler, LabelEncoder
```

## 2. Load Dataset

```python
# Load dataset
df = pd.read_csv('ionosphere.csv')
df.head()
```

## 3. Basic Dataset Information

```python
df.info()
df.describe()
df.isnull().sum()
```

## 4. Handling Missing Values (Imputation)

```python
# Numerical imputation
num_imputer = SimpleImputer(strategy='mean')
df.iloc[:, :-1] = num_imputer.fit_transform(df.iloc[:, :-1])

# Categorical imputation
cat_imputer = SimpleImputer(strategy='most_frequent')
df['class'] = cat_imputer.fit_transform(df[['class']])
```

## 5. Detect Outliers using IQR

```python
Q1 = df.iloc[:, :-1].quantile(0.25)
Q3 = df.iloc[:, :-1].quantile(0.75)
IQR = Q3 - Q1

outliers = ((df.iloc[:, :-1] < (Q1 - 1.5 * IQR)) |
            (df.iloc[:, :-1] > (Q3 + 1.5 * IQR)))

print('Number of outliers:', outliers.sum().sum())
```

## 6. Data Transformations

```python
# Scaling
scaler = StandardScaler()
scaled_features = scaler.fit_transform(df.iloc[:, :-1])
```

```python
# Encoding categorical column
le = LabelEncoder()
df['class_encoded'] = le.fit_transform(df['class'])
df.head()
```

## 7. Feature Engineering

```python
# Feature 1: Mean signal strength
df['signal_mean'] = df.iloc[:, :10].mean(axis=1)

# Feature 2: Signal variance
df['signal_variance'] = df.iloc[:, :10].var(axis=1)

df.head()
```

## Conclusion
This notebook demonstrates data preprocessing steps including missing value handling, outlier detection,
feature scaling, encoding, and feature engineering using ionosphere radar signal data.
